```python
# take the path of the dataset

import kagglehub

# Download latest version
path = kagglehub.dataset_download("maharshipandya/-spotify-tracks-dataset")

print("Path to dataset files:", path)
```

```python
# import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
# check files in the downloaded folder

print(os.listdir(path))
```

```python
# load the CSV file

csv_file = [file for file in os.listdir(path) if file.endswith(".csv")][0]
file_path = os.path.join(path, csv_file)
df = pd.read_csv(file_path)
print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# display column names

print(df.columns)
```

```python
# display first 5 rows

df.head()
```

```python
# display last 5 rows

df.tail()
```

```python
# display random 5 rows

df.sample(5)
```

```python
# dataset information

df.info()
```

```python
# check data types

print(df.dtypes)
```

```python
# check missing values

print(df.isnull().sum())
```

```python
# check duplicate rows

print("Duplicate rows:", df.duplicated().sum())
```

```python
# remove unnecessary column if present

if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])
print(df.shape)
```

```python
# remove duplicate tracks using track_id

print("Duplicate track IDs:", df["track_id"].duplicated().sum())
df = df.drop_duplicates(subset="track_id")
print("Shape after removing duplicates:", df.shape)
```

```python
# handle missing artist names

df["artists"] = df["artists"].fillna("Unknown Artist")
```

```python
# handle missing genre names

df["track_genre"] = df["track_genre"].fillna("Unknown Genre")
```

```python
# handle missing album names

df["album_name"] = df["album_name"].fillna("Unknown Album")
```

```python
# convert duration from milliseconds to minutes

df["duration_minutes"] = df["duration_ms"] / 60000
df[["track_name", "duration_ms", "duration_minutes"]].head()
```

```python
# handle extreme popularity values

df["popularity"] = df["popularity"].clip(0, 100)
print("Minimum popularity:", df["popularity"].min())
print("Maximum popularity:", df["popularity"].max())
```

```python
# final missing value check

print(df.isnull().sum())
```

```python
# select required columns

spotify_df = df[
    [
        "track_name",
        "artists",
        "album_name",
        "track_genre",
        "popularity",
        "duration_minutes",
        "danceability",
        "energy",
        "loudness",
        "speechiness",
        "acousticness"
    ]
].copy()
spotify_df.head()
```

```python
# rename columns

spotify_df = spotify_df.rename(columns={
    "track_name": "Song_Name",
    "artists": "Artist",
    "album_name": "Album",
    "track_genre": "Genre",
    "popularity": "Popularity",
    "duration_minutes": "Duration",
    "danceability": "Danceability",
    "energy": "Energy",
    "loudness": "Loudness",
    "speechiness": "Speechiness",
    "acousticness": "Acousticness"
})
spotify_df.head()
```

```python
# normalize numerical variables

from sklearn.preprocessing import MinMaxScaler
numeric_columns = [
    "Popularity",
    "Duration",
    "Danceability",
    "Energy",
    "Loudness",
    "Speechiness",
    "Acousticness"
]
scaler = MinMaxScaler()
spotify_normalized = spotify_df.copy()
spotify_normalized[numeric_columns] = scaler.fit_transform(spotify_normalized[numeric_columns])
spotify_normalized.head()
```

```python
# top 10 most popular songs

top_songs = spotify_df.sort_values(by="Popularity",ascending=False).head(10)
print(top_songs[["Song_Name", "Artist", "Genre", "Popularity"]])
```

```python
# top artists by average popularity

artist_popularity = (spotify_df.groupby("Artist")["Popularity"].mean().sort_values(ascending=False).head(10))
print(artist_popularity)
```

```python
# plot top artists

plt.figure(figsize=(10, 6))
artist_popularity.sort_values().plot(kind="barh")
plt.title("Top 10 Artists by Average Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Artist")
plt.tight_layout()
plt.show()
```

```python
# duration vs popularity

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=spotify_df,
    x="Duration",
    y="Popularity",
    hue="Popularity",
    palette="viridis"
)
plt.title("Song Duration vs Popularity")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Popularity")
plt.legend(title="Popularity")
plt.show()
```

```python
# correlation between duration and popularity

print("Duration-Popularity Correlation:",spotify_df["Duration"].corr(spotify_df["Popularity"]))
```

```python
# danceability vs popularity

plt.figure(figsize=(10, 6))

spotify_df["Popularity_Level"] = spotify_df["Popularity"].apply(
    lambda x: "Low Popularity" if x < 50 else "High Popularity")
sns.scatterplot(
    data=spotify_df,
    x="Danceability",
    y="Popularity",
    hue="Popularity_Level"
)

plt.title("Danceability vs Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.legend(title="Popularity Level")
plt.show()
```

```python
# correlation

print("Danceability-Popularity Correlation:",spotify_df["Danceability"].corr(spotify_df["Popularity"]))
```

```python
# energy vs popularity

plt.figure(figsize=(10, 6))

spotify_df["Popularity_Level"] = spotify_df["Popularity"].apply(
    lambda x: "Low Popularity" if x < 50 else "High Popularity")

sns.scatterplot(
    data=spotify_df,
    x="Energy",
    y="Popularity",
    hue="Popularity_Level"
)

plt.title("Energy vs Popularity")
plt.xlabel("Energy")
plt.ylabel("Popularity")
plt.legend(title="Popularity Level")
plt.show()
```

```python
# correlation

print("Energy-Popularity Correlation:",spotify_df["Energy"].corr(spotify_df["Popularity"]))
```

```python
# count songs by genre

genre_counts = spotify_df["Genre"].value_counts()
print(genre_counts.head(10))
```

```python
# top 10 dominant genres

top_genres = genre_counts.head(10)
plt.figure(figsize=(10, 6))
top_genres.sort_values().plot(kind="barh")
plt.title("Top 10 Dominant Genres")
plt.xlabel("Number of Songs")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

```python
# average popularity by genre

genre_popularity = (spotify_df.groupby("Genre")["Popularity"].mean().sort_values(ascending=False))
print(genre_popularity.head(10))
```

```python
# top 10 genres by popularity

plt.figure(figsize=(10, 6))
genre_popularity.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Genres by Average Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

```python
# correlation matrix

correlation_columns = [
    "Popularity",
    "Duration",
    "Danceability",
    "Energy",
    "Loudness",
    "Speechiness",
    "Acousticness"
]

correlation_matrix = spotify_df[correlation_columns].corr()
print(correlation_matrix)
```

```python
# correlation heatmap

plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Spotify Audio Features Correlation")
plt.tight_layout()
plt.show()
```

```python
# final dataset information

print("Final Shape:", spotify_df.shape)
print("\nColumns:")
print(spotify_df.columns.tolist())

print("\nMissing Values:")
print(spotify_df.isnull().sum())

print("\nDuplicate Rows:")
print(spotify_df.duplicated().sum())
```

```python
# save cleaned dataset

spotify_df.to_csv("spotify_tracks_cleaned.csv",index=False)
print("Cleaned dataset saved successfully!")
```
